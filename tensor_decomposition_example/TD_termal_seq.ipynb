```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.decomposition import FactorAnalysis, PCA
# import tensortools as tt
# from tensortools.operations import unfold as tt_unfold, khatri_rao
import tensorly as tl
from tensorly import unfold as tl_unfold
from tensorly.decomposition import parafac
import struct
import time
import os

from utils import *
```

```python
path = 'B004_1_1_frames.npz'
file_size = os.path.getsize(path)

print(f"File size: {file_size} bytes")
```

```python
frames_data = np.load(path)
frames_data = frames_data['arr_0']
print(frames_data.shape)

no_of_frames = frames_data.shape[0]
height = frames_data.shape[1]
width = frames_data.shape[2]
```

```python
img_data = frames_data.reshape(no_of_frames,height,width)
plt.imshow(img_data[1000], cmap='gray')
plt.show()
```

```python
# Specify the tensor and the rank
X, rank = img_data[:,:,:], 3
```

```python
# Perform CP decompositon using TensorLy
factors_tl = parafac(X, rank=rank)
print(factors_tl)
```

```python
# Reconstruct M, with the result of each library
M_tl = reconstruct(factors_tl.factors)
```

```python
# Compute the reconstruction error
rec_error_tl = np.mean((X-M_tl)**2)
```

```python
# plot the decomposed factors from TensorLy
plot_factors(factors_tl.factors, d=2)
plt.suptitle("Factors computed with TensorLy", y=1.1, fontsize=20);
```
